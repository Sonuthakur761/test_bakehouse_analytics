```python
def palindrome(text:str):
    ch=''
    for word in text:
        ch=word+ch
    if ch==text:
        return True
    return False 

if palindrome('vijay'):
    print('Palindrome')
else:
    print('Not Palindrome')   
```

```python
palindrome('madam')
```

```python
def str_word_count(text:str):
    d={}
    for i in text:
        if i in d:
            d[i]=d[i]+1
        else:
            d[i]=1
    return d        
```

```python
str_word_count('vijay pratap singh')
```

```python
def str_word_count_2(text:str):
    d={}
    for i in text:
        d[i]=d.get(i,0)+1
    return d 
```

```python
str_word_count_2('vijjjsy')
```
